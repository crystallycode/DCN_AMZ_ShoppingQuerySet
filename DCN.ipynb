```python
from torch.utils.data import Dataset, DataLoader
class CustomDataset(Dataset):
    def __init__(
            self, data, 
            embedding_model_name="bert-base-cased", 
            locale_le=None,
            esci_le=None
        ):
        super().__init__()

        self.data_df = data
        self.tokenizer = AutoTokenizer.from_pretrained(embedding_model_name)
        self.embedding_model = AutoModel.from_pretrained(embedding_model_name)
        self.embedding_model.eval()
        for param in self.embedding_model.parameters():
            param.requires_grad = False
        print(f"BERT embedding model '{embedding_model_name}' loaded, frozen, and set to eval mode in Dataset.")

        self.locale_le = locale_le
        self.esci_le = esci_le

        self.device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        self.embedding_model.to(self.device)

    def __len__(self):
        return len(self.data_df)
    
    def _get_text_embedding(self, texts):
        # print(f"Getting text embedding for: {texts}")
        if not texts:
            texts = ""
        if isinstance(texts, float) and np.isnan(texts):
            texts = ""
        if not isinstance(texts, list):
            texts = [texts]
        with torch.no_grad():
            tokens = self.tokenizer(texts, padding=True, truncation=True, return_tensors='pt')
            tokens = {k: v.to(self.device) for k, v in tokens.items()} # Move input tokens to the same device
            outputs = self.embedding_model(**tokens)
            return outputs.last_hidden_state[:, 0, :]


    def __getitem__(self, idx):
        item = self.data_df.iloc[idx]

        query = self._get_text_embedding(item['query']).squeeze(0)

        product_title = self._get_text_embedding(item['product_title']).squeeze(0)
        # print(f"Product title tensor: {product_title}")
        product_description = self._get_text_embedding(item['product_description']).squeeze(0)
        product_bullet_point = self._get_text_embedding(item['product_bullet_point']).squeeze(0)
        product_brand = self._get_text_embedding(item['product_brand']).squeeze(0)
        product_color = self._get_text_embedding(item['product_color']).squeeze(0)

        product_locale = self.locale_le.transform([item['product_locale']])
        # print(product_locale)
        product_locale = torch.tensor(product_locale, dtype=torch.long)
        # print(f"Product locale tensor: {product_locale}")

        esci_label = self.esci_le.transform([item['esci_label']])
        esci_label = torch.tensor(esci_label, dtype=torch.long)

        return {
            'query_embed': query,
            'product_title_embed': product_title,
            'product_description_embed': product_description,
            'product_bullet_point_embed': product_bullet_point,
            'product_brand_embed': product_brand,
            'product_color_embed': product_color,
            'product_locale': product_locale
        }, esci_label
```

```python
class DCNWithEmbeddings(nn.Module):
    def __init__(
            self, 
            text_embedding_dim, 
            num_text=6,
            num_classes=4, 
            locale_vocab_size=3, 
            locale_emb_dim=8,
            deep_hidden_dims: list = [512, 256, 128], # Dimensions of deep layers
            dropout_rate: float = 0.2,
            apply_cross_net: bool = True,
            cross_net_type: str = 'parallel', # 'parallel' or 'stack'
            # freeze_embedding_model: bool = True
        ):
        super().__init__()

        self.text_embedding_dim = text_embedding_dim
        
        self.locale_emb = nn.Embedding(locale_vocab_size, locale_emb_dim)
        # self.cross = ... # your cross layers
        total_input_dim = num_text * self.text_embedding_dim + locale_emb_dim
        # deep layer
        deep_layers = []
        current_input_dim = total_input_dim
        for hidden_dim in deep_hidden_dims:
            deep_layers.append(nn.Linear(current_input_dim, hidden_dim))
            deep_layers.append(nn.ReLU())
            deep_layers.append(nn.Dropout(dropout_rate))
            current_input_dim = hidden_dim
        self.deep = nn.Sequential(*deep_layers)
        if apply_cross_net and cross_net_type == 'stack':
            self.cross_net = CrossNet(current_input_dim, layer_num=2, parameterization='vector')
        if apply_cross_net and cross_net_type == 'parallel':
            parallel_cross_net_input_dim = num_text * self.text_embedding_dim + locale_emb_dim
            self.cross_net = CrossNet(parallel_cross_net_input_dim, layer_num=2, parameterization='vector')
            self.parallel_cross_net_head = nn.Sequential(
                nn.Linear(parallel_cross_net_input_dim + deep_hidden_dims[-1], deep_hidden_dims[-1]),
                nn.ReLU(),
                nn.Dropout(dropout_rate)
            )

        self.classifier = nn.Linear(current_input_dim, num_classes)

        self.apply_cross_net = apply_cross_net
        self.cross_net_type = cross_net_type


    def forward(self, batch):
        # batch: dict with keys for each field
        query_emb = batch['query_embed']
        # print("Query embedding shape:", query_emb.shape)
        title_emb = batch['product_title_embed']
        # print("Title embedding shape:", title_emb.shape)
        desc_emb = batch['product_description_embed']
        # print("Description embedding shape:", desc_emb.shape)
        bullet_emb = batch['product_bullet_point_embed']
        # print("Bullet point embedding shape:", bullet_emb.shape)
        brand_emb = batch['product_brand_embed']
        # print("Brand embedding shape:", brand_emb.shape)
        color_emb = batch['product_color_embed']
        # print("Color embedding shape:", color_emb.shape)
        locale_emb = self.locale_emb(batch['product_locale'].squeeze(-1))
        # print("Product locale tensor shape:", batch['product_locale'].squeeze(-1).shape)
        # print("Locale embedding shape:", locale_emb.shape)
        # print("Locale embedding:", locale_emb)
        # Concatenate all embeddings
        x = torch.cat([query_emb, title_emb, desc_emb, bullet_emb, brand_emb, color_emb, locale_emb], dim=1)
        # Pass through DN layers and classifier
        if not self.apply_cross_net:
            x = self.deep(x)
            logits = self.classifier(x)
        if self.apply_cross_net and self.cross_net_type == 'stack':
            x = self.deep(x)
            x = self.cross_net(x)
            logits = self.classifier(x)
        if self.apply_cross_net and self.cross_net_type == 'parallel':
            deep_x = self.deep(x)
            cross_x = self.cross_net(x)
            x = torch.cat([deep_x, cross_x], dim=1)
            x = self.parallel_cross_net_head(x)
            logits = self.classifier(x)            

        
        return logits
```

```python
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

NUM_CLASSES = len(esci_le.classes_) # Example: 4
LOCALE_VOCAB_SIZE = len(locale_le.classes_) # Example: 4
LOCALE_EMB_DIM = 8
TEXT_EMBEDDING_DIM = 768 # For bert-base-uncased
NUM_TEXT = 6
DEEP_HIDDEN_DIMS = [512, 256, 128]
DROPOUT_RATE = 0.2



model = DCNWithEmbeddings(
    text_embedding_dim = TEXT_EMBEDDING_DIM, 
    num_text = NUM_TEXT,
    num_classes = NUM_CLASSES, 
    locale_vocab_size = LOCALE_VOCAB_SIZE, 
    locale_emb_dim = LOCALE_EMB_DIM,
    deep_hidden_dims = DEEP_HIDDEN_DIMS, # Dimensions of deep layers
    dropout_rate = DROPOUT_RATE,
)

model.to(DEVICE)

model
```

```python
train_dataset = CustomDataset(
    data=df_examples_products[df_examples_products['split'] == 'train'],
    locale_le=locale_le,
    esci_le=esci_le,
    embedding_model_name="bert-base-cased"
)
test_dataset = CustomDataset(
    data=df_examples_products[df_examples_products['split'] == 'test'],
    locale_le=locale_le,
    esci_le=esci_le,
    embedding_model_name="bert-base-cased" # Make sure to pass model name
)

train_dataloader = DataLoader(train_dataset, batch_size=16, shuffle=True, num_workers=0) # num_workers=0 for simplicity
test_dataloader = DataLoader(test_dataset, batch_size=16, shuffle=False, num_workers=0) # For GPU data transfer demo
```

```python
def count_trainable_params(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)

print(f"Trainable parameters: {count_trainable_params(model):,}")
```

```python
optimizer = torch.optim.AdamW(model.parameters(), lr=2e-5)
criterion = nn.CrossEntropyLoss()
```

```python
print(f"\nTraining model on: {DEVICE}")
print("Starting training loop...")
```

```python
num_epochs = 1 # Try
for epoch in range(num_epochs):
    model.train()
    total_loss = 0
    for batch_idx, (inputs, labels) in enumerate(train_dataloader):
        # terminate after 11 batches
        if batch_idx > 10:
            break

        # --- HERE IS WHERE YOU MOVE BATCH DATA TO GPU ---
        # `inputs` is the dictionary from your Dataset, `labels` is the target tensor
        
        # Move all tensors within the 'inputs' dictionary to the device
        for key in inputs:
            inputs[key] = inputs[key].to(DEVICE)
            # print(f"Device of {key}: {inputs[key].device}")

            
        # Move the labels tensor to the device
        labels = labels.squeeze(-1)  
        labels = labels.to(DEVICE)
        # ----------------------------------------------------

        # --- DEBUGGING LINES ---
        # print(inputs)
        # print(f"Device of locale_emb.weight: {model.locale_emb.weight.device}")

        optimizer.zero_grad()
        for key in inputs:
            inputs[key] = inputs[key].to(DEVICE)
            # print(f"Device of input: {inputs[key].device}")
        
        logits = model(inputs) # Pass the dictionary of tensors
        logits = logits.to(DEVICE)
        # print(f"Logits: {logits}, Labels: {labels}")
        # print(f"Logits shape: {logits.shape}, Labels shape: {labels.shape}")
        loss = criterion(logits, labels)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()

        if batch_idx % 10 == 0:
            print(f"Epoch {epoch+1}, Batch {batch_idx}, Loss: {loss.item():.4f}")

    avg_train_loss = total_loss / len(train_dataloader)
    print(f"Epoch {epoch+1} finished. Average Train Loss: {avg_train_loss:.4f}")

    # (Optional) Validation loop
    model.eval()
    test_loss = 0
    with torch.no_grad():
        for i, (inputs, labels) in enumerate(test_dataloader):
            # terminate after 100 batches for test set
            if i > 11:
                break
            # Move validation data to GPU as well
            for key in inputs:
                inputs[key] = inputs[key].to(DEVICE)
            
            labels = labels.squeeze(-1)  
            labels = labels.to(DEVICE)

            logits = model(inputs)
            loss = criterion(logits, labels)
            test_loss += loss.item()

            if i % 10 == 0:
                print(f"Test set: Epoch {epoch+1}, Batch {i}, Loss: {loss.item():.4f}")

    avg_test_loss = test_loss / len(test_dataloader)
    print(f"Test set Loss: {avg_test_loss:.4f}")

print("Training complete!")
```

### Evaluation metrics

```python
# ESCI classification
import torch
from sklearn.metrics import classification_report
import numpy as np

model.eval()
val_loss = 0
all_labels = []
all_predictions = []
with torch.no_grad():
    for i, (inputs, labels) in enumerate(test_dataloader):
        if i > 10:
            break
        # Move validation data to GPU as well
        for key in inputs:
            inputs[key] = inputs[key].to(DEVICE)
        
        labels = labels.squeeze(-1)
        labels = labels.to(DEVICE)

        logits = model(inputs)
        logits = logits.to(DEVICE)
        loss = criterion(logits, labels)
        val_loss += loss.item()

        probabilities = torch.softmax(logits, dim=1) # Apply softmax along the class dimension
        predictions = torch.argmax(probabilities, dim=1) # Get the index of the highest probability
        
        # Store labels and predictions
        all_labels.extend(labels.cpu().numpy())
        all_predictions.extend(predictions.cpu().numpy())
avg_val_loss = val_loss / len(test_dataloader)
print(f"Validation Loss: {avg_test_loss:.4f}")


# Generate the classification report
# Convert lists to numpy arrays
y_true = np.array(all_labels)
y_pred = np.array(all_predictions)

print("\n--- Multiclass Classification Report ---")
print(classification_report(y_true, y_pred, target_names=['E', 'S', 'C', 'I']))
```

```python
# NDCG
# Convert label to relevance score and calculate NDCG
from sklearn.metrics import ndcg_score
import numpy as np

# Define mapping from label index to gain
label2gain = np.array([0, 0.01, 0.1, 1])  # I=0, C=1, S=2, E=3 (assuming label encoding order)

def labels_to_relevance(labels):
    # labels: tensor or numpy array of shape (batch,)
    return label2gain[labels.cpu().numpy()]

# Example: calculate NDCG for a batch (use after a batch in validation loop)
# Replace 'labels' and 'logits' with your actual batch variables
if 'labels' in locals() and 'logits' in locals():
    # Get predicted class indices
    pred_labels = logits.argmax(dim=1)
    # Convert to relevance scores
    true_relevance = labels_to_relevance(labels)
    pred_relevance = labels_to_relevance(pred_labels)
    # NDCG expects shape (1, n_samples)
    ndcg = ndcg_score(true_relevance.reshape(1, -1), pred_relevance.reshape(1, -1))
    print(f"NDCG for this batch: {ndcg:.4f}")
else:
    print("Run this cell after a batch to see NDCG.")
```
